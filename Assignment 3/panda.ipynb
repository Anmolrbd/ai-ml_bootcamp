### Series : It is a pandas' 1-Dimensional Labaled Array that can hold any data type. 

Analogy: single column in a spreedsheet.
They can be created from List, Tuple, Dictionary, Numpy array, and Scaler value


```python
## Creating a List series
# room_number = [20,30,40]
# room_series = pd.Series(room_number, index = ["appartment #1", "apparment #2", "apparment #3"]) ##custom index=[...]



data1 = [100, 200, 300]

series1 = pd.Series(data1,index = ["a","b","c"]) ##Here Series() is a constructor not a funtion.
print(series1) ##arranges the data into a column in a labeled manner

##location by label
print(series1.loc["a"]) #accessing/locating the value via label
# print(series1.loc["d"]) ##does not have index d, hence returns key index error

#chainging the value of a series.
series1.loc["c"] = 400
print(series1)

##locating the value by integer position
print(series1.iloc[1]) ##iloc: interger location
```

### Filtering

```python
data2 = [10, 20, 30, 40, 50, 60]

series2 = pd.Series(data2, index = ["a","b","c","d","e","f"])


print(series2[series2 >= 40])
```

## Creating a series from a Dictionary

```python
calories = {
    "Day 1" : 2500,
    "Day 2" : 2400,
    "Day 3" : 2800,
}

calories_series = pd.Series(calories)
print(calories_series)

print(calories_series.loc["Day 2"]) #accessing 
print()
calories_series.loc["Day 2"] += 500 #chaning the data 
print()
print(calories_series[calories_series > 2500])
```

```python
#creating a random List
pokemon = ["Bulbasaurt","ivysaur","Venusaur","Charmander","Charmeleon","Charizard"]

numberlist = [val for val in range(1,7)]
pokemonseries = pd.Series(pokemon, index = numberlist)
print(pokemonseries)
```

# Dataframe: A tabular structure with rows and columns.(2D)
## Analogy: Similar to an excel spreedsheet

```python
company = {
    "Name": ["SpongeBob", "Patrick","Squidward"],
    "Age" : [30, 35, 50]
}

df1 = pd.DataFrame(company, index=["Employee 1","Employee 2","Employee 3"]) ##constructor 
print(df1)
print()
print(df1.loc["Employee 1"]) #prints the entire row of the specific label
print(df1.iloc[2]) #prints the entire row using index location
```

```python
#adding a new column
df1["Job"] = ["Cook","N/A","Cashier"]
print(df1)
```

```python
#adding a new row


#recreating again because in Juypter df holds the value and keeps on adding the value in the same df
df1 = pd.DataFrame({
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, 30, 22],
    "Job": ["Doctor", "Teacher", "Designer"]
}, index=["Employee 1", "Employee 2", "Employee 3"])

new_row = pd.DataFrame(
    [{"Name": "Sandy", "Age": 28, "Job": "Engineer"},
     {"Name": "Larry", "Age": 38, "Job": "Bodybuilder"}],
    index=["Employee 4", "Employee 5"]
)

df1 = pd.concat([df1, new_row])

print(df1)
```
