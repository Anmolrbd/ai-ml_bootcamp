```python

import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
```

## Importing files

```python

iris = pd.read_csv('iris.csv')
stocks = pd.read_csv('stocks-monthly.csv',parse_dates=[0])
```

## 1. Scatterplot

```python

#using plt
plt.scatter(iris['SepalLengthCm'], iris['SepalWidthCm'])

plt.xlabel('SepalWidthCm') #naming the X axis

plt.ylabel('SepalLengthCm') #naming the X axis

plt.title("Scatterplot") #naming the X axis

#add gridlines
plt.grid()
```

```python
#using sns
sns.scatterplot(
    x=iris['SepalWidthCm'],
    y=iris['SepalLengthCm'],
    hue=iris['Species']
).set(title="Scatterplot")
```

## Barplot

```python

#using the stocks.csv file

plt.bar(stocks['Date'].dt.month, stocks['GOOG']) #"Create a bar chart where the x-axis contains months from the Date column, and the y-axis contains Google's stock prices."
#dt.month selects the month column

#add legend
plt.legend(['GOOG'])

#name the X axis
plt.xlabel('Month')

#name the Y axis
plt.ylabel('Closing price')

#name the plot
plt.title("Bar plot")

#add gridlines
plt.grid()
```

```python
#using seaborn


sns.barplot(
    x=stocks['Date'].dt.month,
    y=stocks['GOOG']
).set(title="Bar plot")
```

## Stack Bar chart

```python



plt.bar(stocks['Date'].dt.month, stocks['GOOG'], color='r')
plt.bar(stocks['Date'].dt.month, stocks['MSFT'], bottom=stocks['GOOG'], color='b')
plt.bar(stocks['Date'].dt.month, stocks['AMZN'], bottom=stocks['GOOG']+stocks['MSFT'], color='y')
plt.bar(stocks['Date'].dt.month, stocks['NASDAQ'], bottom=stocks['GOOG']+stocks['MSFT']+stocks['AMZN'], color='g')

#name the X axis
plt.xlabel("Months")

#name the Y axis
plt.ylabel("Closing price")

#add legend
plt.legend(["GOOG", "MSFT", "AMZN", "NASDAQ"])

#add title
plt.title("Closing prices of stocks in each month")

#add limit for Y axis to better visualize all categories
plt.ylim(0,800)
```

```python

sns.lineplot(data = stocks, x = 'Date', y = 'NASDAQ').set(title="Line plot")
```
