## Lists

```python
lists=[1,2,3,"dog","cat","buffalo",0,9]
print(lists)
lists.append("banana")
print(lists)

alist=[val **2 for val in range(10)]
print(alist)
```

## Dictionary

```python
colors={"cat":"brown",
        "dog":"black",
        "speak":["meow","woof"]}

for key,value in colors.items():
  print(key,value)
```

## Creating a Tuple

```python
# Creating an empty tuple
empty_tuple = ()

print(empty_tuple)
print(type(empty_tuple))
```

```python
# Creating a tuple with values

numbers = (1, 2, 3, 4, 5)

print(numbers)
print(type(numbers))
```

```python
# Tuple with different data types

student = ("Anmol", 19, "Computer Engineering", 8.5)

print(student)
```

## Accessing Tuple Elements

```python
colors = ("red", "green", "blue", "yellow")

# First element
print(colors[0])

# Second element
print(colors[1])

# Last element
print(colors[-1])

#slicing
numbers = (10,20,30,40,50)

# Elements from index 1 to 3
print(numbers[1:4])
# From beginning to index 2

print(numbers[:3])
# From index 2 to end

print(numbers[2:])
```

## Loops through tuples

```python
fruits = ("apple","banana","orange")

for fruit in fruits:
    print(fruit)
```
