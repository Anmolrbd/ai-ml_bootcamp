```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

sumed_value = np.sum(sales_data[:,1:], axis=0)
print(np.sum(sumed_value))

#minimun sales per restaurant
min_sales = np.min(sales_data[:,1:],axis=1)
print(min_sales)

max_sales = np.max(sales_data[:,1:],axis=0)
print(max_sales)

average_sales = np.average(sales_data[:,1:],axis=1)
print(average_sales)


cumsum = np.cumsum(sales_data[:,1:], axis=1)
print(cumsum)


plt.figure(figsize= (8,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cumulative sales")
plt.xlabel("Years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

vector_addition = vector1 + vector2
print("Addition:", vector_addition)

dotproduct = np.dot(vector1, vector2)
print("Dot product: ", dotproduct)

scalemul = vector1 * 3
print("Multiplied by 3",scalemul)

elementmult = vector1 * vector2
print("Element wise multiplication: ", elementmult)

v1magnitude = np.linalg.norm(vector1)
print("Magnitude of vector1: ", v1magnitude) 
```
